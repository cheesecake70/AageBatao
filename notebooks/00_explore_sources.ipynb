# Explore sources
Length and Code-Mixing Index distributions per source.

```python
import pandas as pd, matplotlib.pyplot as plt
D='../data/hingmix/v0.1/'
f={k:pd.read_parquet(D+k+'.parquet',columns=['tokens','cmi','n_switch']) for k in ['train','val','test_a','test_b']}
for d in f.values(): d['length']=d.tokens.str.len()
pd.DataFrame({k:d[['length','cmi','n_switch']].mean() for k,d in f.items()}).round(2)
```

```python
fig,ax=plt.subplots(1,2,figsize=(11,4))
for k,d in f.items():
    ax[0].hist(d.length,bins=range(3,62,3),density=True,histtype='step',label=k)
    ax[1].hist(d.cmi,bins=range(0,55,5),density=True,histtype='step',label=k)
ax[0].set_xlabel('tokens per message'); ax[1].set_xlabel('CMI'); ax[0].legend(); plt.show()
```
